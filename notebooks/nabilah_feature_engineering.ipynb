```python
to_impute_ph = ["weight_lbs", "length_in", "width_in", "height_in"]
for colname in to_impute_ph:
    df[colname + '_na'] = df[colname].isnull()
for colname in to_impute_ph:
    df[colname] = df[colname].fillna(np.mean(df[colname]))
for colname in to_impute_ph:
    print('Missing values: ' + str(colname) + ': ' + str(np.sum(df.columns.isnull())) )
```

```python
suspicious = df[(df["weight_lbs"]>150) & (df["pallet_count"]==0)].copy()
suspicious["converted_weight"] = suspicious["weight_lbs"]/453.592
suspicious[["converted_weight", "weight_lbs", "length_in", "width_in", "height_in", "pallet_count"]].sort_values("weight_lbs", ascending=False).head(100)
```

```python
condition = (df["weight_lbs"]>150) & (df["pallet_count"]==0)
df["weight_lbs"] = np.where(condition, df["weight_lbs"]/453.592, df["weight_lbs"])
df[["weight_lbs", "length_in", "width_in", "height_in", "pallet_count"]].sort_values("weight_lbs", ascending=False).head(1000)
```
